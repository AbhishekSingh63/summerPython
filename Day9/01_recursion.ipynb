```python
# factorial
def fact(n):
    if n==1:
        return 1
    return n*fact(n-1)
print(fact(4))
```

```python
#reverse countdown
def countdown(n):
    if n==0:
        return
    print(n)
    countdown(n-1)
countdown(5)
```

```python
#print 1st n numbers
def nNum(n):
    if n==0:
        return
    nNum(n-1)
    print(n)
nNum(5)
```

```python
# sum of n numbers
def sumn(n):
    if n==1:
        return 1
    return n+sumn(n-1)
print(sumn(5))
```

```python
#Recursive list sum
def list_sum(arr, n):
    if n == 0:
        return 0
    return arr[n - 1] + list_sum(arr, n - 1)
arr = [10, 20, 30, 40, 50]
result = list_sum(arr, len(arr))
print("Sum =", result)
```

```python
# function returning function
def calculator(operation):
    if operation=='sum':
        return add(3,5)
    elif operation=='subtract':
        return minus(5,2)
def add(a,b):
    return a+b
def minus(a,b):
    return a-b
op=calculator('sum')
print(op)
```

```python
# build your own map
def my_map(func, data):
    result = []
    for i in data:
        result.append(func(i))
    return result
def square(x):
    return x * x
numbers = [1, 2, 3, 4, 5]
output = my_map(square, numbers)
print(output)
```

```python
# recursive string compression
def compress(s, i=0, count=1):
    if i == len(s) - 1:
        return s[i] + str(count)
    if s[i] == s[i + 1]:
        return compress(s, i + 1, count + 1)
    return s[i] + str(count) + compress(s, i + 1, 1)
text = "aaabbbccccdda"
print(compress(text))
```

```python
# lambda + filter
students=[("Alice", 90, 20),("Bob", 80, 21),("Charlie", 88, 19),("David", 92, 22),("Eva", 86, 20)]
result = list(filter(lambda x: x[1] > 85 and x[2] > 19, students))
print(result)
```
