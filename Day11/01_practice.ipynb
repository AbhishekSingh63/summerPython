```python
# find the frequency of each character in a string
s = input("Enter a string: ")
checked = ""
for ch in s:
    if ch not in checked:
        print(ch, ":", s.count(ch))
        checked += ch
```

```python
# covert a snake_case string into CamelCase
s = input("Enter snake_case string: ")
words = s.split("_")
camel = ""
for word in words:
    camel += word.capitalize()
print("CamelCase:", camel)
```

```python
#convert customer names into title case
name = input("Enter customer name: ")
print("Title Case:", name.title())
```

```python
#extract the domain name from an email address
email = input("Enter email address: ")
domain = email.split("@")[1]
print("Domain:", domain)
```

```python
#count the number of letters,digits,spaces, and special characters
s = input("Enter a string: ")
letters = 0
digits = 0
spaces = 0
special = 0
for ch in s:
    if ch.isalpha():
        letters += 1
    elif ch.isdigit():
        digits += 1
    elif ch.isspace():
        spaces += 1
    else:
        special += 1
print("Letters:", letters)
print("Digits:", digits)
print("Spaces:", spaces)
print("Special Characters:", special)
```

```python
#check whether two strings are anagram
s1 = input("Enter first string: ")
s2 = input("Enter second string: ")
if sorted(s1) == sorted(s2):
    print("Anagram")
else:
    print("Not an Anagram")
```

```python
#find longest word in a sentence
sentence = input("Enter a sentence: ")
words = sentence.split()
longest = words[0]
for word in words:
    if len(word) > len(longest):
        longest = word
print("Longest word:", longest)
```
