```python
# Q1 Netflix content Dashboard. Show provided list. display first, last, top 3, last 3, alternate shows, reverse using slicing
content=["avengers","it","got","from","dark"]
print(content)
print(content[0])
print(content[len(content)-1])
print(content[:3])
print(content[-3:])
print(content[::2])
print(content[::-1])
```

```python
# Q2 Hospital patient monitoring
hrate=[90,80,105,95,102]
for i in range(len(hrate)):
    if hrate[i]>100:
        print("patent ",i+1," with high heart rate = ",hrate[i])
```

```python
# Q3 display highest, lowest, total, and average salary
salaries = [35000, 42000, 50000, 28000, 60000]
highest = max(salaries)
lowest = min(salaries)
total = sum(salaries)
average = total / len(salaries)
print("Salaries:", salaries)
print("Highest Salary:", highest)
print("Lowest Salary:", lowest)
print("Total Salary:", total)
print("Average Salary:", average)
```

```python
# Q4 Concept: Membership. Take employee id as input. Print access granted or access denied.
emp_ids = [101, 102, 103, 104, 105]
emp_id = int(input("Enter Employee ID: "))
if emp_id in emp_ids:
    print("Access Granted")
else:
    print("Access Denied")
```
