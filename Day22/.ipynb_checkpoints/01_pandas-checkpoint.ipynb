```python
import pandas as pd
```

```python
# series from a list
marks=pd.Series([85,90,78,92])
print(marks)
```

```python
# series from a dictionary
city_population=pd.Series({'delhi':338000000,'mumbai':'478000000','pune':294788000})
print(city_population)
```

```python
# series from nd array
import numpy as np
arr=np.array([10,20,30])
s=pd.Series(arr)
print(s)
```

```python
# custom index
marks=pd.Series([85,90,75],index=['Math','Science','English'])
print(marks)
```

```python
# series attributes
s=pd.Series([85,90,78,92], name='Marks')
print(s)
print(s.values)
print(s.index)
print(s.dtype)
print(s.shape)
print(s.size)
print(s.name)
```

```python
# series methods
s=pd.Series([85,90,78,92], name='Marks')
print(s)
print(s.head(2))
print(s.tail(2))
print(s.sum())
print(s.mean())
print(s.max())
print(s.min())
print(s.sort_values())
print(s.unique())
print(s.nunique())
print(s.apply(lambda x: x * 2))
```

```python
# from a dictionary
data={'name':['aditi','rohan','meera'],
      'age':[23,21,19],
      'city':['lucknow','delhi','jaipur']}
df=pd.DataFrame(data)
print(df)
```

```python
# from a list of dictionary
records=[
    {'name':'aditi','age':21},
    {'name':'rohan','age':23}
]
df=pd.DataFrame(records)
print(df)
```

```python
# using nested list
rows=[['aditi',21],['rohan',23]]
df=pd.DataFrame(rows,columns=['Name','Age'])
print(df)
```

```python
# from a numpy array
arr=[[43,21],[25,23]]
df=pd.DataFrame(arr,columns=['N','A'])
print(df)
```

```python
rows=[['aditi',21],['rohan',23]]
df=pd.DataFrame(rows,columns=['Name','Age'])
print(df)
print(df.head(1))
print(df.tail(1))
print(df.sample(2))
print(df.info())
print(df.describe())
print(df.shape)
print(df.columns)
print(df.index)
print(df.dtypes)
print(df['Age'].value_counts())
print(df['Age'].unique())
print(df['Age'].nunique())
```

```python
# data cleaning
```

```python
# missing values
rows=[['aditi',21],['rohan',23],['',23],['ricky',]]
df=pd.DataFrame(rows,columns=['Name','Age'])
print(df.isnull())
print(df.isnull().sum())
print(df.notnull())
```

```python
# dropna() fillna()
print(df.dropna())
print(df.dropna(subset=['Age']))
print(df.dropna(how='all'))

print(df['Age'].fillna(df['Age'].mean()))
print(df['Age'].fillna(df['Age'].ffill())) # forward fill from the previous row
```

```python
# replace(), rename(), duplicate rows

import pandas as pd

data = {
    'StudentID': [101, 102, 103, 104, 104, 105],
    'Name': ['Rahul', 'Priya', 'Amit', 'Sneha', 'Sneha', 'Rahul'],
    'Age': [20, 21, 19, 22, 22, 23],
    'gender': ['M', 'F', 'M', 'F', 'F', 'M'],
    'Marks': [85, 92, 78, 88, 88, 90],
    'City':['delhi','pune','lucknow','jaipur','mumbai','assam']
}
df = pd.DataFrame(data)

print(df['gender'].replace({'M':'Male', 'F':'Female'}))

print(df.rename(columns={'StudentID':'Student_ID'}))

print(df.duplicated()) # boolean mask of duplicate rows
print(df.drop_duplicates()) # remove duplicate rows
print(df.drop_duplicates(subset=['Name'], keep='first'))
```

```python
# astype()
df['Age']=df['Age'].astype(int)
```

```python
# string cleaning
df['Name']=df['Name'].str.strip()
# etc
```
