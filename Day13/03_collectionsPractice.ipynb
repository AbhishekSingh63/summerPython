```python
# Q1. store the last 15 tempratures recorded during a day and display the highest lowest and average temprature
temps = []
for i in range(15):
    temp = float(input("Temperature " + str(i + 1) + ": "))
    temps.append(temp)
print("Highest Temperature:", max(temps))
print("Lowest Temperature:", min(temps))
print("Average Temperature:", sum(temps) / len(temps))
```

```python
# Q2. store the unique programming languages known by a class and store them alphabetically
languages = set()
for i in range(5):
    lang = input("Enter programming language: ")
    languages.add(lang)
print(sorted(languages))
```
