```python
# Q1 accept 5 string values from user and store them as part of tuple. print the elements of tuple with there indices.
items = []
for i in range(5):
    value = input("Enter a string: ")
    items.append(value)
t = tuple(items)
for i in range(len(t)):
    print(i, ":", t[i])
```

```python
# Q2 Accept 5 string values from user and store them as part of tuple. 
#Print the elements of tuple with there indices in reverse order.
items2 = []
for i in range(5):
    value = input("Enter a string: ")
    items2.append(value)
t2 = tuple(items2)
t3=reversed(t2)
index = len(t2) - 1
for item in (t3):
    print(index, ":", item)
    index -= 1
```
