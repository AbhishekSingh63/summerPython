```python
# set methods- add(),update(),remove(),discard(),pop(),copy(),clear()
s={10,20,30}
s.add(50)
print(s)
s.update([50,60])
print(s)
s.remove(20)
s.discard(100)
print(s)
print(s.pop())
copy=s.copy()
copy.clear()
print(copy)
```

```python
# adding
student["city"]="delhi"

# updating
student["age"]=23
student.update({"cgpa":8.7})
print(student)

# remove
student.pop("city")
print(student)

student.popitem() # remove last key value pair
print(student)

student.clear()
print(student)
```

```python
# methods- get(), keys(), values(), items(), update(), pop(), popitem(), clear(), copy(), setdefault(), fromkeys().
# nested dictionary
students = {
    101: {"name": "Rahul", "cgpa": 8.1},
    102: {"name": "Priya", "cgpa": 8.9}
}
print(students[101]["name"])
print(students[102]["cgpa"])
```

```python
# dictionary comprehension
square={x:x*x for x in range(1,11)}
print(square)
evensquare={x:x*x for x in range(1,21) if x%2==0}
print(evensquare)
```

```python
# 1.create a student dictionary 2.add a new key 3.update an existing value 4.delete a key using pop()
#5.print all keys 6.print all values 7.traverse using items() 8.create a nested dictionary 
#9.create a dictionary comprehension for cubes 10.count frequency of characters in a string using a dictionary

#1
student = {
    "Name": "Rahul",
    "Age": 20,
    "Branch": "CSE"
}
print(student)

#2
student["City"] = "Delhi"
print(student)

#3
student["Age"] = 21
print(student)

#4
student.pop("City")
print(student)

#5
print(student.keys())

#6
print(student.values())

#7
for key, value in student.items():
    print(key, ":", value)

#8
students = {
    101: {"Name": "Rahul", "Age": 20},
    102: {"Name": "Priya", "Age": 21}
}
print(students)

#9
cubes = {x: x**3 for x in range(1, 6)}
print(cubes)

#10
s = input("Enter a string: ")
freq = {}
for ch in s:
    freq[ch] = freq.get(ch, 0) + 1
print(freq)
```
