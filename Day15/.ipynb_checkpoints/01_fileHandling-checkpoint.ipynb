```python
''' Create a file and write your introduction. Read and display file contents. Append a new line.
Count number of lines. Count number of words. Search a word. Copy one file into another.
Convert file content to uppercase. Store 5 student names. Read ulternate lines. '''

# Create and write introduction
with open("intro.txt", "x") as file:
    pass
print("File created successfully.")

with open("intro.txt", "w") as file:
    file.write("My name is Abhishek.\n")
    file.write("I am a BTech CSE student.\n")
    file.write("I study at Lovely Professional University.\n")

# Read file
with open("intro.txt", "r") as file:
    print(file.read())

# Append
with open("intro.txt", "a") as file:
    file.write("I love Python programming.\n")

# Count number of lines
with open("intro.txt", "r") as file:
    lines = file.readlines()
    print("Number of lines:", len(lines))

# Count number of words
with open("intro.txt", "r") as file:
    words = file.read().split()
    print("Number of words:", len(words))

# Search a word
word = input("Enter word to search: ")
with open("intro.txt", "r") as file:
    content = file.read()
if word in content:
    print(word, "found in file.")
else:
    print(word, "not found.")

# Copy one file into another
with open("intro.txt", "r") as source:
    data = source.read()

with open("copy.txt", "x") as file:
    pass
print("Copy file created successfully.")

with open("copy.txt", "w") as destination:
    destination.write(data)
print("File copied successfully.")

# Convert file content to uppercase
with open("intro.txt", "r") as file:
    upper = file.read().upper()
print(upper)

# Store 5 student names
with open("student.txt", "w") as file:
    file.write("Aman\n")
    file.write("Rahul\n")
    file.write("Abhishek\n")
    file.write("Priya\n")
    file.write("Neha\n")

# Read alternate lines
with open("student.txt", "r") as file:
    lines = file.readlines()
for i in range(0, len(lines), 2):
    print(lines[i], end="")
```
