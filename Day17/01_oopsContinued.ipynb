```python
# create a base class person with name and age. create teacher and student subclasses, each adding one extra attribute (e.g. subject for teacher, 
#roll_no for student) and overriding a common method role() to describe who they are.
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

class Teacher(Person):
    def __init__(self, name, age, subject):
        super().__init__(name, age)
        self.subject = subject
    def role(self):
        print("I am a Teacher")

class Student(Person):
    def __init__(self, name, age, roll_no):
        super().__init__(name, age)
        self.roll_no = roll_no
    def role(self):
        print("I am a Student")

t1 = Teacher("Rahul", 35, "Mathematics")
s1 = Student("Abhishek", 20, 101)

t1.role()
s1.role()
```

```python
# create square and triangle classes, each with a draw() method that prints a different simple pattern (e.g. using *). Put objects of both in a list 
#and loop through, calling draw() on each.
class Square:
    def draw(self):
        print("Square:")
        for i in range(4):
            print("* * * *")
        print()

class Triangle:
    def draw(self):
        print("Triangle:")
        for i in range(1, 5):
            print("* " * i)
        print()

s = Square()
t = Triangle()
shapes = [s, t]

for shape in shapes:
    shape.draw()
```

```python
# Smart Library System: Design a system to issue and return books. Store book availability safely, allow members to borrow/return books, 
#and calculate late fees if applicable.
class Book:
    def __init__(self, title):
        self.title = title
        self.__available = True

    def is_available(self):
        return self.__available

    def issue(self):
        if self.__available:
            self.__available = False
            return True
        return False

    def return_book(self):
        self.__available = True

class Member:
    def __init__(self, name):
        self.name = name
        self.borrowed_books = []

    def borrow_book(self, book):
        if book.issue():
            self.borrowed_books.append(book)
            print(self.name, "borrowed", book.title)
        else:
            print(book.title, "is not available")

    def return_book(self, book, late_days):
        if book in self.borrowed_books:
            self.borrowed_books.remove(book)
            book.return_book()
            print(self.name, "returned", book.title)

            fee = late_days * 5
            print("Late Fee =", fee)
        else:
            print("This book was not borrowed by", self.name)

class Library:
    def __init__(self):
        self.books = []

    def add_book(self, book):
        self.books.append(book)

    def show_books(self):
        print("Library Books")
        for book in self.books:
            if book.is_available():
                print(book.title, "- Available")
            else:
                print(book.title, "- Issued")

library = Library()

b1 = Book("Python")
b2 = Book("Data Structures")
b3 = Book("Machine Learning")

library.add_book(b1)
library.add_book(b2)
library.add_book(b3)

m1 = Member("Abhishek")

library.show_books()

print()

m1.borrow_book(b1)
m1.borrow_book(b2)

print()

library.show_books()

print()

m1.return_book(b1, 3)

print()

library.show_books()
```
