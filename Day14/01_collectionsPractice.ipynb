```python
# remove duplicate roll numbers while preserving the original order
roll = [9,10,2,1,4,4,5,2]
unique = []
for i in roll:
    if i not in unique:
        unique.append(i)
print(unique)
```

```python
# given marks of students, print names of students scoring above class average
marks = {"Rahul": 85,"Amit": 72,"Priya": 90,"Neha": 65,"Rohan": 78}
average = sum(marks.values()) / len(marks)
print("Average Marks: ",average)
for name, score in marks.items():
    if score > average:
        print(name)
```

```python
# create an airline reservation model and detect double-booked seats.
reservations = {"FL101": ["A1", "A2", "A3", "A2"],
                "FL102": ["B1", "B2", "B3"]
               }
for flight in reservations:
    seats = reservations[flight]
    if len(seats) != len(set(seats)):
        print(flight, "has double-booked seats")
    else:
        print(flight, "has no double booking")
```
