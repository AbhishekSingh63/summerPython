```python
# create a book class with attributes title, author, price. add a method display_info() that prints all details.
class Book:
    def __init__(self, title, author, price):
        self.title = title
        self.author = author
        self.price = price
    def display_info(self):
        print("Title :", self.title)
        print("Author:", self.author)
        print("Price :", self.price)
        print()
b1 = Book("Python Programming", "Abhishek Singh", 499)
b1.display_info()
```

```python
# create a student class with roll_no, name, marks. add a method get_marks() that returns "A" if marks >=90, "B" if >=75, else "c"
class Student:
    def __init__(self, roll_no, name, marks):
        self.roll_no = roll_no
        self.name = name
        self.marks = marks
    def get_marks(self):
        if self.marks >= 90:
            return "A"
        elif self.marks >= 75:
            return "B"
        else:
            return "C"
s1 = Student(101, "Abhishek", 88)
print("Roll No :", s1.roll_no)
print("Name    :", s1.name)
print("Marks   :", s1.marks)
print("Grade   :", s1.get_marks())
```

```python
# create a rectangle class with legth and width. add methods area() and perimeter()
class Rectangle:
    def __init__(self,length,width):
        self.length=length
        self.width=width
    def area(self):
        print(self.length*self.width)
    def perimeter(self):
        print(2*(self.length+self.width))
rec1=Rectangle(10,20)
rec1.area()
rec1.perimeter()
```

```python
# create a employee class with name and salary. add a method give_raise(percent) that updates self.salary in place 
#(no return, just modify and print new salary).
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
    def give_raise(self, percent):
        self.salary = self.salary + (self.salary * percent / 100)
        print("Updated Salary =", self.salary)

emp1 = Employee("Abhishek", 50000)
print("Current Salary =", emp1.salary)
emp1.give_raise(10)
```

```python
# (challenge - ties to earlier collections topic). create a library class that holds a list of Book objects (from Q1). Add methods add_book(book) and show_all_books().
class Library:
    def __init__(self):
        self.books = []
        
    def add_book(self, book):
        self.books.append(book)
        
    def show_all_books(self):
        print("Books in Library:")
        for book in self.books:
            book.display_info()

b1 = Book("Python Programming", "Abhishek Singh", 499)
b2 = Book("Data Structures", "Rahul Sharma", 650)
b3 = Book("Machine Learning", "Amit Verma", 799)

lib = Library()

lib.add_book(b1)
lib.add_book(b2)
lib.add_book(b3)

lib.show_all_books()
```

```python
# Q.Encapsulation: Create an Employee class with a private attribute __salary. Add a getter and a setter - the setter should reject negative salaries.
#Add a method apply_raise(percent) that increases salary using the setter internally.
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.__salary = salary
    # Getter
    def get_salary(self):
        return self.__salary
    # Setter
    def set_salary(self, salary):
        if salary >= 0:
            self.__salary = salary
        else:
            print("Salary cannot be negative.")

    def apply_raise(self, percent):
        new_salary = self.__salary + (self.__salary * percent / 100)
        self.set_salary(new_salary)
        print("Updated Salary =", self.__salary)

emp = Employee("Abhishek", 50000)
print("Employee Name :", emp.name)
print("Current Salary:", emp.get_salary())

# Applying 10% raise
emp.apply_raise(10)
```
