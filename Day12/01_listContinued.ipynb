```python
# Mini project. Library Management System: Store books, add books, search, issue, return 
#,create backup, sort books and display final library using appropriate list methods.
books = ["Python", "Java", "C++", "DBMS"]
print(books)

# Add a book
new_book = input("Enter book to add: ")
books.append(new_book)
print("Book Added Successfully!")

# Search a book
search = input("Enter book to search: ")
if search in books:
    print("Book Found")
else:
    print("Book Not Found")
    
# Issue a book
issue = input("Enter book to issue: ")
if issue in books:
    books.remove(issue)
    print("Book Issued Successfully!")
else:
    print("Book Not Available")
    
# Return a book
returned = input("Enter book to return: ")
books.append(returned)
print("Book Returned Successfully!")

# Create Backup
backup = books.copy()
print("Library Backup:")
print(backup)

# Sort Books
books.sort()

# Display Final Library
print("Final Library:")
print(books)
```
