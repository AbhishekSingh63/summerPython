```python
# sum
sumLambda = lambda x,y: x+y
print(sumLambda(5,8))
# multiply
multiplyLambda = lambda x,y: x*y
print(multiplyLambda(5,8))
```

```python
# largest number
from functools import reduce
numbers = [4, 17, 92, 8, 55]
largest = reduce(lambda x, y: x if x > y else y, numbers)
print(largest)
```

```python
# check prime
is_prime_lambda = lambda n: n > 1 and all(n % i != 0 for i in range(2, int(n**0.5) + 1))
print(is_prime_lambda(17))
print(is_prime_lambda(15))
```

```python
# celcius to fahrenhiet
ctof = lambda x: (x*9/5)+32
print(ctof(10))
```

```python
# multiplication tables
table=lambda n: [n * i for i in range(1, 11)]
print(table(4))
```

```python
#cube
num=[5,6,3,9]
cube=list(map(lambda x:x**3,num))
print(cube)
```

```python
# even odd
evenodd=lambda n: "even" if n%2==0 else "odd"
print(evenodd(7))
```

```python
# positive negative
possneg=lambda n: "positive" if n>=0 else "negative"
print(possneg(-7))
```

```python
# apply 20% discount
price=[500,650,350,590]
disPrice=list(map(lambda x:x-(x*20/100),price))
print(disPrice)
```
