```python
#Q1
def area(l,b):
    return l*b
leng=int(input("Enter length of table"))
bre=int(input("Enter breadth of table"))
a=area(leng,bre)
print("area is: ",a)
```

```python
#Q2
def interest(p,r,t):
    return (p*r*t)/100
p=int(input("Enter principal"))
r=int(input("Enter rate"))
t=int(input("Enter time period"))
si=interest(p,r,t)
print("Interest is: ",si)
```

```python
#Q3
def tofaren(cel):
    return (cel*9/5)+32
cel=int(input("temp in celcius"))
far=tofaren(cel)
print("Temp. in farenhite: ", far)
```

```python
#Q4
#gst=12%, 8%
def calcGst(total):
    if(total>10000):
        return total*12/100;
    elif(total<=10000):
        return total*8/100;
a=int(input("total without gst: "))
gst=calcGst(a)
final=a+gst
print("total sum after gst: ", final)
```

```python
#Q5
def eligible(age):
    if age>=18:
        print("Eligible")
    else:
        print("Not Eligible")
age=int(input("enter candidate age: "))
eligible(age)
```

```python
#Q6
def auth(name, passw):
    if 4 < len(name) < 18 and name[0] != '@' and 5 < len(passw) < 12:
        print("Valid User")
    else:
        print("Not a Valid User")
name = input("Enter username: ")
passw = input("Enter password: ")
auth(name, passw)
```

```python
#Q7
def family_member(**details):
    print("Family Member Details:")
    for key, value in details.items():
        print(f"{key} : {value}")
name=input("name: ")
age=int(input("age: "))
city=input("city: ")
occupation=input("occupation: ")
mobile=input("mobile: ")
family_member(name=name,age=age,city=city,occupation=occupation,mobile=mobile)
```

```python
#Q8
def generate_bill(price, quantity, discount=0):
    total = price * quantity
    final_amount = total - discount
    return final_amount
price = float(input("Enter price of one item: "))
quantity = int(input("Enter quantity: "))
discount = float(input("Enter discount: ")) # if no discount then 0
bill = generate_bill(price, quantity, discount)
print("Final Payable Amount =", bill)
```

```python
#Q9
def student_result(name, *marks):
    total = sum(marks)
    average = total / len(marks)
    print("Total Marks:", total)
    print("Average Marks:", average)
    if average >= 40:
        print("Result: Pass")
    else:
        print("Result: Fail")
        
name = input("Enter student name: ")
n = int(input("Enter number of subject: "))
marks = []
for i in range(n):
    mark = int(input("Enter marks of subject: "))
    marks.append(mark)
student_result(name, *marks)
```

```python
#Q10
def calculate_salary(basicsal, hra, da):
    finalsal = basicsal + hra + da
    return finalsal
basicsal = float(input("Enter Basic Salary: "))
hra = float(input("Enter HRA: "))
da = float(input("Enter DA: "))
salary = calculate_salary(basicsal, hra, da)
print("Final Salary :", salary)
```
