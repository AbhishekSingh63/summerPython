```python
#Q2 calculator
def calculator(num1,num2,op):
    if op=='+':
        return num1+num2
    elif op=='-':
        return num1-num2
    elif op=='*':
        return num1*num2
    elif op=='//':
        return num1//num2
    elif op=='/':
        return num1/num2
    elif op=='%':
        return num1%num2
    elif op=='**':
        return num1**num2
n1=int(input())
n2=int(input())
o=input()
calculator(n1,n2,o)
```

```python
#Q3 largest of two
def largest(num1,num2):
    if num1>num2:
        print("Larger number is ",num1)
    elif num2>num1:
        print("Larger number is ",num2)
    else:
        print("Both are equal") 
num1=int(input())
num2=int(input())
largest(num1,num2)
```

```python
#Q4 even or odd
def evenodd(num1):
    if num1%2==0:
        print("Even")
    else:
        print("Odd") 
num1=int(input())
evenodd(num1)
```

```python
#Q5 largest of 5 numbers
def largest(*nums):
    max=nums[0]
    for i in nums:
        if i>max:
            max=i
    return max
result=largest(2,5,3,8,5,11)
print("largest number is ",result)
```
