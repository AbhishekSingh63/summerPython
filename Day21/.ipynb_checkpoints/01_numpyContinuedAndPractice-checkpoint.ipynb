```python
import numpy as np
```

```python
'''
Student Marks analysis System using only Numpy -
1-dataset creation
2-average marks per student
3-highest and lowest marks
4-standard deviation per student
5-ranking students
6-top students(above class average)
7-subject wise-analysis
8-percentage calculation
9-vizualization-ready data (this step reshapes results into simple (label,value) pairs ready to handover to matplotlib or pandas for charting.)
'''
```

```python
students=["Rahul","Priya","Amit","Sneha","Rohan"]

subjects=["Maths","Science","English","Computer"]

marks=np.array([
    [85, 90, 78, 92],
    [70, 88, 84, 76],
    [95, 91, 89, 94],
    [60, 72, 68, 65],
    [88, 79, 92, 81]
])
```

```python
average_marks=np.mean(marks, axis=1)
print("Average Marks of each Student")
for i in range(len(students)):
    print(students[i], ":", average_marks[i])
```

```python
highest_marks=np.max(marks, axis=1)
lowest_marks=np.min(marks, axis=1)
print("Highest and Lowest Marks of each Student")
for i in range(len(students)):
    print(students[i],highest_marks[i],lowest_marks[i])
```

```python
std_marks=np.std(marks, axis=1)
print("Standard Deviation of each Student")
for i in range(len(students)):
    print(students[i],":",std_marks[i])
```

```python
total_marks=np.sum(marks, axis=1)
ranking=np.argsort(total_marks)[::-1]
print("Student Ranking:")
rank = 1
for i in ranking:
    print(rank,":",students[i],"-",total_marks[i])
    rank += 1
```

```python
average_marks=np.mean(marks, axis=1)
class_avg=np.mean(average_marks)
print("Class Average: ",class_avg)
print("Students Above Average")
for i in range(len(students)):
    if average_marks[i]>class_avg:
        print(students[i], ":", average_marks[i])
```

```python
subject_average = np.mean(marks, axis=0)
print("Subject-wise Average")
for i in range(len(subjects)):
    print(subjects[i], ":", subject_average[i])
```

```python
total_marks = np.sum(marks, axis=1)
percentage = (total_marks / 400) * 100
print("Student Percentage")
for i in range(len(students)):
    print(students[i], ":", round(percentage[i], 2))
```
